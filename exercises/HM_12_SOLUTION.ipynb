## שלב 1 - קבצים בסיסיים

ממשו `missing_basics(files)`: מחזירה **רשימה** של קודי ממצא עבור ארבעה קבצים בסיסיים שחסרים: `"no_readme"`, `"no_license"`, `"no_tests"`, `"no_ai_log"`.

**כלים:** `list(files)` נותן את **שמות** הקבצים. `any(תנאי for n in names)` מחזיר `True` אם **לפחות אחד** מהשמות מקיים את התנאי. `n.lower()` ו-`n.upper()` מתעלמים מאותיות גדולות/קטנות. `n.startswith("tests/")` בודק התחלה.

**שימו לב:** קובץ בדיקות הוא `test_...` או בתיקיית `tests/`. הקובץ `latest_report.py` **אינו** קובץ בדיקות, גם אם יש בו את האותיות `test`.

</div>

````python
def missing_basics(files):
    """files: dict {file name: text}. Returns a list of finding codes (strings), each at most once:
    "no_readme"  - no file named README.md (capitalization does not matter)
    "no_license" - no file whose name starts with LICENSE (LICENSE, LICENSE.txt, ...)
    "no_tests"   - no file starting with "tests/" and no file starting with "test_"
    "no_ai_log"  - no file named AI_LOG.md (capitalization does not matter)
    e.g. missing_basics({"app.py": ""}) -> ["no_readme", "no_license", "no_tests", "no_ai_log"]
    """
    names = list(files)
    findings = []
    if not any(n.lower() == "readme.md" for n in names):
        findings.append("no_readme")
    if not any(n.upper().startswith("LICENSE") for n in names):
        findings.append("no_license")
    if not any(n.startswith("tests/") or n.startswith("test_") for n in names):
        findings.append("no_tests")
    if not any(n.lower() == "ai_log.md" for n in names):
        findings.append("no_ai_log")
    return findings
````

````python
# בדיקה - חלק א', שלב 1
def check_a1():
    assert sorted(missing_basics({"app.py": ""})) == ["no_ai_log", "no_license", "no_readme", "no_tests"], \
        "missing_basics: בפרויקט עם app.py בלבד חסרים כל ארבעת הקבצים"
    assert missing_basics(store_good) == [], "missing_basics: ב-store_good יש את כל הקבצים, צריך להחזיר רשימה ריקה"
    assert missing_basics({"readme.md": "", "LICENSE.txt": "", "ai_log.md": "", "test_a.py": ""}) == [], \
        "missing_basics: אותיות קטנות (readme.md), LICENSE.txt ו-test_a.py צריכים להיחשב"
    assert "no_tests" in missing_basics(store_messy), \
        "missing_basics: latest_report.py אינו קובץ בדיקות - 'test' בתוך מילה אחרת (latest) לא נחשב. בדקו התחלה של שם: startswith"
    assert missing_basics({"tests/a.py": "", "README.md": "", "LICENSE": "", "AI_LOG.md": ""}) == [], \
        "missing_basics: קובץ בתוך תיקיית tests/ נחשב בדיקה"
    return True

check_a1() and print("✓ שלב 1: הקבצים הבסיסיים נבדקים")
````

## שלב 2 - מפתח שנשכח בקוד

הסעיף החשוב ביותר באבטחה: **אין מפתח או סיסמה בקוד או בהיסטוריה של Git** (מפגש 10). ממשו `find_secrets(files)` שמחזירה **רשימה של שמות הקבצים החשודים** (לא את הסוד עצמו: לא מדפיסים סודות!).

קובץ חשוד אם **אחד** מאלה מתקיים:

- **השם** (החלק האחרון אחרי ה-`/` האחרון) הוא בדיוק `.env`, או מסתיים ב-`.pem` (מפתח פרטי). שימו לב: `.env.example` **אינו** חשוד - הוא התבנית בלי ערכים, והיא כן אמורה לעלות ל-Git.
- **התוכן** מכיל תבנית של מפתח. התבנית כבר כתובה: `SECRET_PATTERN` (ביטוי רגולרי), ובודקים אותה כך: `SECRET_PATTERN.search(text)` מחזיר התאמה, או `None` אם אין.

**כלים:** `name.split("/")[-1]` נותן את החלק האחרון של הנתיב. `name.endswith(".pem")` בודק סיום.

</div>

````python
# תבנית (ביטוי רגולרי) לזיהוי מפתחות - כתובה בשבילכם, אין צורך לשנות
SECRET_PATTERN = re.compile(r"""(\bsk-[A-Za-z0-9]{16,}|AKIA[0-9A-Z]{16}|api[_-]?key\s*[=:]\s*["'][^"'\s]{8,}["'])""", re.IGNORECASE)

print(bool(SECRET_PATTERN.search('KEY = "sk-FAKE0123456789abcdefXYZ"')), bool(SECRET_PATTERN.search("KEY = os.environ.get('LLM_KEY')")))
````

````python
def find_secrets(files):
    """Returns the list of file NAMES that look like they contain a secret (never the secret itself).
    A file is suspicious if its last name part is exactly ".env" or ends with ".pem",
    or if SECRET_PATTERN.search(its text) finds a match.
    e.g. find_secrets({".env.example": "K=", "config/.env": "K=1"}) -> ["config/.env"]
    """
    suspicious = []
    for name, text in files.items():
        last = name.split("/")[-1]
        if last == ".env" or last.endswith(".pem") or SECRET_PATTERN.search(text):
            suspicious.append(name)
    return suspicious
````

````python
# בדיקה - חלק א', שלב 2
def check_a2():
    assert find_secrets(store_good) == [], "find_secrets: ב-store_good אין סוד. שימו לב: .env.example אינו .env"
    assert find_secrets(store_leaky) == ["app.py"], "find_secrets: ב-store_leaky המפתח כתוב בתוך app.py (בדקו את התוכן, לא רק את השם)"
    assert find_secrets(store_messy) == [".env"], "find_secrets: ב-store_messy הקובץ .env עצמו חשוד"
    assert find_secrets({"config/.env": "A=1"}) == ["config/.env"], "find_secrets: .env בתוך תיקייה (config/.env) - בדקו את החלק האחרון של השם"
    assert find_secrets({"key.pem": ""}) == ["key.pem"], "find_secrets: קובץ שמסתיים ב-.pem חשוד"
    assert find_secrets({"x.py": 'api_key = "abc12345xyz"'}) == ["x.py"], "find_secrets: api_key עם ערך בתוך הקוד חשוד"
    assert find_secrets({"task.md": "ask-for-help"}) == [], "find_secrets: טקסט רגיל אינו מפתח"
    return True

check_a2() and print("✓ שלב 2: מפתחות נתפסים לפי שם ולפי תוכן")
````

## שלב 3 - קישור שבור (מבנית)

בעמוד נחיתה ובקובץ README יש קישורים. ב-Colab **אין לנו רשת** כדי לבדוק אם אתר באמת עונה, אבל אפשר לתפוס קישורים ש**שבורים כבר במבנה** שלהם. ממשו `is_broken_link(link, files, folder="")` שמחזירה `True` אם הקישור שבור. הכללים, **לפי הסדר** (הראשון שמתקיים מכריע):

1. הקישור ריק, או רק `#` (אחרי `link.strip()`) ← **שבור**.
2. הקישור מכיל `example.com`, `localhost`, `127.0.0.1` או `TODO` (באותיות גדולות) ← **שבור**: זה placeholder או כתובת מקומית.
3. הקישור מתחיל ב-`#` (עוגן בתוך העמוד, כמו `#demo`), או ב-`http://`, `https://`, `mailto:` ← **תקין**. (לא בודקים שהאתר עונה: את זה בודקים בדפדפן.)
4. אחרת זה קישור **יחסי** לקובץ: שבור אם הקובץ **לא במאגר**. הנתיב המלא הוא `folder + link` (בלי החלק אחרי `#`, אם יש), ו-`folder` הוא התיקייה של הקובץ שבו נמצא הקישור, למשל `"docs/"`.

**כלים:** `link.strip()`; `"x" in link`; `link.startswith(("#", "http://"))` - אפשר לתת **סוגריים עם כמה אפשרויות** (tuple); `link.split("#")[0]` לוקח את מה שלפני `#`; `path not in files` בודק אם נתיב חסר במילון.

הפונקציות `find_links`, `folder_of` ו-`all_links` כבר כתובות: הן מוצאות את כל הקישורים בקבצי `.md` ו-`.html` (קישור ב-HTML: `href="..."`, ב-Markdown: `[טקסט](קישור)`).

</div>

````python
# כתוב בשבילכם - הריצו וקראו
LINK_RE = re.compile(r"""href=["']([^"']*)["']|\]\(([^)]*)\)""")

def find_links(text):
    """All link targets in a text: href="..." (HTML) and [text](...) (Markdown)."""
    return [a or b for a, b in LINK_RE.findall(text)]

def folder_of(name):
    """'docs/index.html' -> 'docs/' ;  'README.md' -> ''"""
    return name.rsplit("/", 1)[0] + "/" if "/" in name else ""

def all_links(files):
    """Every link in every .md / .html file: a list of (link, folder of the file that contains it)."""
    pairs = []
    for name, text in files.items():
        if name.endswith((".md", ".html")):
            pairs += [(link, folder_of(name)) for link in find_links(text)]
    return pairs

print(all_links(store_good))
````

````python
def is_broken_link(link, files, folder=""):
    """True if the link is broken by its structure. Rules in order:
    1. empty or just "#"  -> True
    2. contains example.com / localhost / 127.0.0.1 / TODO  -> True
    3. starts with "#", "http://", "https://" or "mailto:"  -> False
    4. otherwise a file link: True if (folder + link without the part after "#") is not in files
    e.g. is_broken_link("style.css", {"docs/style.css": ""}, "docs/") -> False
    """
    link = link.strip()
    if link in ("", "#"):
        return True
    if "example.com" in link or "localhost" in link or "127.0.0.1" in link or "TODO" in link:
        return True
    if link.startswith(("#", "http://", "https://", "mailto:")):
        return False
    return folder + link.split("#")[0] not in files
````

````python
# בדיקה - חלק א', שלב 3
def check_a3():
    f = {"docs/style.css": "", "docs/page.html": "", "a.md": ""}
    assert is_broken_link("", f) and is_broken_link("  ", f) and is_broken_link("#", f), "is_broken_link: קישור ריק או # בלבד הוא שבור"
    assert is_broken_link("https://example.com", f), "is_broken_link: example.com הוא placeholder - שבור"
    assert is_broken_link("http://localhost:8501", f), "is_broken_link: localhost הוא כתובת מקומית - שבור"
    assert is_broken_link("TODO", f), "is_broken_link: TODO הוא placeholder - שבור"
    assert not is_broken_link("#demo", f), "is_broken_link: עוגן #demo בתוך העמוד תקין"
    assert not is_broken_link("https://the-store.test/demo", f), "is_broken_link: כתובת https רגילה תקינה (לא בודקים את הרשת)"
    assert not is_broken_link("https://todoist.com/x", f), "is_broken_link: TODO באותיות גדולות בלבד (todoist אינו placeholder)"
    assert not is_broken_link("mailto:a@b.test", f), "is_broken_link: mailto תקין"
    assert not is_broken_link("style.css", f, "docs/"), "is_broken_link: style.css בתיקייה docs/ קיים - הנתיב המלא הוא folder + link"
    assert is_broken_link("style.css", f, ""), "is_broken_link: בלי folder הנתיב הוא style.css, והוא לא במאגר"
    assert is_broken_link("old.html", f, "docs/"), "is_broken_link: docs/old.html לא קיים"
    assert not is_broken_link("page.html#top", f, "docs/"), "is_broken_link: מתעלמים מהחלק אחרי #"
    return True

check_a3() and print("✓ שלב 3: קישורים שבורים נתפסים")
````

## שלב 4 - הביקורת המלאה

חברו הכול ל-`prelaunch_audit(files)`: מחזירה **רשימת ממצאים** (קודים, כל קוד פעם אחת), בשלושה צעדים:

1. התחילו מ-`missing_basics(files)`.
2. אם `find_secrets(files)` **לא ריקה** - הוסיפו `"secret_committed"`.
3. אם **לפחות קישור אחד** מ-`all_links(files)` שבור - הוסיפו `"broken_link"`. **רמז:** `any(is_broken_link(l, files, f) for l, f in all_links(files))`. (כל איבר ב-`all_links` הוא זוג: קישור ותיקייה.)

</div>

````python
def prelaunch_audit(files):
    """Returns a list of finding codes for a project (each code at most once; order does not matter):
    the codes of missing_basics, plus "secret_committed" and "broken_link" (see the table above).
    An empty list means: no finding.
    """
    findings = missing_basics(files)
    if find_secrets(files):
        findings.append("secret_committed")
    if any(is_broken_link(l, files, f) for l, f in all_links(files)):
        findings.append("broken_link")
    return findings
````

````python
# בדיקה - חלק א', שלב 4
def check_a4():
    assert prelaunch_audit(store_good) == [], "prelaunch_audit: store_good מושלם - רשימה ריקה"
    assert sorted(prelaunch_audit(store_almost)) == ["broken_link", "no_ai_log"], \
        "prelaunch_audit: ב-store_almost חסר AI_LOG.md, וה-README מקשר אליו (קישור שבור)"
    assert prelaunch_audit(store_leaky) == ["secret_committed"], "prelaunch_audit: ב-store_leaky רק המפתח ב-app.py"
    got = prelaunch_audit(store_messy)
    assert sorted(got) == ["broken_link", "no_ai_log", "no_license", "no_readme", "no_tests", "secret_committed"], \
        "prelaunch_audit: ב-store_messy יש כל ששת הממצאים"
    assert len(got) == len(set(got)), "prelaunch_audit: כל קוד אמור להופיע פעם אחת (גם אם יש כמה קישורים שבורים)"
    return True

check_a4() and print("✓ שלב 4: הביקורת עובדת על ארבעה פרויקטים")
````

## שלב 5 - מותר לחשוף?

לא כל ממצא עוצר חשיפה. ההגדרה במעבדה: **חוסם** (`"block"`) = חשיפה או נזק מיידיים לאחרים (סוד שדלף, אין רישיון, אין README, אין בדיקות). **אזהרה** (`"warn"`) = חייבים לסגור עד מפגש 13, אבל לא עוצרת את העלייה לאוויר. הטבלה `SEVERITY` כבר כתובה. (ההחלטה מה חוסם היא החלטה של הצוות; אפשר לשנות אותה בבונוס.)

ממשו `ready_to_launch(findings)`: מחזירה `True` רק אם **אף אחד** מהממצאים אינו `"block"`.

**כלים:** `SEVERITY[code]` נותן את החומרה. `any(...)` / `all(...)` על רשימה.

</div>

````python
SEVERITY = {"secret_committed": "block", "no_license": "block", "no_readme": "block", "no_tests": "block",
            "no_ai_log": "warn", "broken_link": "warn"}

def ready_to_launch(findings):
    """findings: list of codes. True only if NONE of them is "block" in SEVERITY.
    e.g. ready_to_launch(["no_ai_log"]) -> True ;  ready_to_launch(["no_tests"]) -> False
    """
    return not any(SEVERITY[code] == "block" for code in findings)
````

````python
# בדיקה - חלק א', שלב 5
def check_a5():
    assert ready_to_launch([]) is True, "ready_to_launch: בלי ממצאים - מוכן"
    assert ready_to_launch(["no_ai_log", "broken_link"]) is True, "ready_to_launch: אזהרות בלבד לא עוצרות חשיפה"
    assert ready_to_launch(["no_tests"]) is False, "ready_to_launch: no_tests הוא חוסם"
    assert ready_to_launch(["no_ai_log", "secret_committed"]) is False, "ready_to_launch: מספיק חוסם אחד, גם אם יש אזהרות"
    return True

check_a5() and print("✓ שלב 5: ההחלטה לחשיפה עובדת")
````

<div dir="rtl">

**לפני שמריצים:** מבין חמשת הפרויקטים (`store_good`, `store_almost`, `store_leaky`, `store_messy`, `store_hollow`) - אילו יסומנו "מוכן לחשיפה"? ולגבי `store_hollow` (יש בו את כל הקבצים, אבל README הוא `TODO` והבדיקה היא `assert True`): מה תחזיר הביקורת, ומה דעתכם על התשובה?

**התחזית שלי:** מוכנים: good (בלי ממצאים), almost (שתי אזהרות בלבד) וגם hollow, כי יש בו את כל הקבצים. leaky ו-messy לא מוכנים (סוד, ובנוסף חוסמים אחרים ב-messy). התשובה על hollow מטעה: הביקורת רואה שיש ראיה, לא שהיא שווה משהו.

</div>

````python
def print_report(name, files):
    findings = prelaunch_audit(files)
    blocking = [f for f in findings if SEVERITY[f] == "block"]
    warnings = [f for f in findings if SEVERITY[f] == "warn"]
    print(f"{name}:  חוסמים: {blocking}  אזהרות: {warnings}  ->  מוכן לחשיפה: {'כן' if ready_to_launch(findings) else 'לא'}")
    if "secret_committed" in findings:
        print("      קבצים חשודים:", find_secrets(files))

for name, files in PROJECTS.items():
    print_report(name, files)
````

## שלב 1 - קישור, או placeholder

ממשו `link_or_placeholder(url, text, placeholder)`:

- אם `url` מתחיל ב-`http://` או `https://` ← מחזירה את הטקסט `<a class="button" href="URL">TEXT</a>` (ה-URL והטקסט **אחרי `html.escape`**).
- אחרת (ריק, או כל דבר אחר, כמו `javascript:alert(1)`) ← מחזירה `<span class="note">PLACEHOLDER</span>` (גם הוא אחרי `html.escape`), **בלי קישור בכלל**.

**למה escape?** כל טקסט שמגיע מבחוץ (מהמשתמש, מ-AI, מקובץ) הוא **נתונים, לא קוד** (מפגש 10). אם שם הפרויקט הוא `A&B <i>`, בלי escape הוא ישבור את ה-HTML או יזריק תגיות.

**כלים:** `html.escape(text)` מחליפה `<`, `>`, `&` וגם מרכאות בקודים בטוחים (חשוב בתוך `href="..."`). בניית טקסט עם משתנים: `f'<a href="{x}">{y}</a>'`. `url.startswith(("http://", "https://"))`.

</div>

````python
def link_or_placeholder(url, text, placeholder):
    """Returns '<a class="button" href="URL">TEXT</a>' when url starts with http:// or https://,
    otherwise '<span class="note">PLACEHOLDER</span>'. Everything is passed through html.escape.
    e.g. link_or_placeholder("", "demo", "coming soon") -> '<span class="note">coming soon</span>'
    """
    if url.startswith(("http://", "https://")):
        return f'<a class="button" href="{html.escape(url, quote=True)}">{html.escape(text)}</a>'
    return f'<span class="note">{html.escape(placeholder)}</span>'
````

````python
# בדיקה - חלק ב', שלב 1
def check_b1():
    out = link_or_placeholder("https://a.test/x", "לחצו", "בקרוב")
    assert 'href="https://a.test/x"' in out and "לחצו" in out and out.startswith("<a"), "link_or_placeholder: כתובת https תקינה צריכה להפוך ל-<a ... href=...>טקסט</a>"
    out = link_or_placeholder("", "לחצו", "כאן יופיע סרטון")
    assert "<a" not in out and "href" not in out and "כאן יופיע סרטון" in out, "link_or_placeholder: בלי כתובת - רק placeholder, בלי קישור (קישור ריק הוא באג)"
    out = link_or_placeholder("javascript:alert(1)", "לחצו", "בקרוב")
    assert "javascript" not in out and "<a" not in out, "link_or_placeholder: כתובת שאינה http/https (כמו javascript:) לא הופכת לקישור"
    out = link_or_placeholder('https://a.test/?q="x"', "t", "p")
    assert "&quot;" in out, "link_or_placeholder: מרכאות בתוך הכתובת צריכות escape (html.escape(url, quote=True))"
    out = link_or_placeholder("https://a.test", "<b>t</b>", "p")
    assert "&lt;b&gt;" in out and "<b>" not in out, "link_or_placeholder: גם הטקסט עובר html.escape"
    out = link_or_placeholder("", "t", "A&B")
    assert "A&amp;B" in out, "link_or_placeholder: גם ה-placeholder עובר html.escape"
    return True

check_b1() and print("✓ שלב 1: קישור או placeholder")
````

## שלב 2 - בונים את העמוד

ממשו `build_landing(project)`: מחזירה את **כל העמוד** (מחרוזת HTML) לפי `TEMPLATE`. הפונקציה `li(items)` (למטה) כבר כתובה: היא הופכת רשימה של טקסטים ל-`<li>...</li><li>...</li>`, עם escape.

**מתכון:** בנו מילון `values` שבו כל מפתח הוא שם מקום-להחלפה (בלי הסוגריים) והערך הוא ה-HTML שיוכנס, ואז:

```python
page = TEMPLATE
for key, value in values.items():
    page = page.replace("{{" + key + "}}", value)
return page
```

מה הערכים:

| מפתח | הערך |
|---|---|
| `name`, `tagline`, `workflow`, `ai_disclosure` | הטקסט מ-`project`, אחרי `html.escape` |
| `video` | `link_or_placeholder(project["video_url"], "▶ צפו בסרטון ההדגמה", "כאן יופיע סרטון ההדגמה")` |
| `demo` | `link_or_placeholder(project["demo_url"], "לדמו החי", "הדמו יתווסף בהמשך")` |
| `steps`, `limits` | `li(...)` של הרשימה המתאימה |

</div>

````python
def li(items):
    """['a', 'b'] -> '<li>a</li><li>b</li>' (each text is escaped). Written for you."""
    return "".join(f"<li>{html.escape(item)}</li>" for item in items)

print(li(["קלט", "A&B"]))
````

````python
def build_landing(project):
    """project: dict with keys name, tagline, video_url, demo_url, steps (list), workflow, ai_disclosure, limits (list).
    Returns the full HTML page (a string) built from TEMPLATE. Free texts go through html.escape;
    a missing video / demo link stays a placeholder (never an empty link).
    """
    values = {
        "name": html.escape(project["name"]),
        "tagline": html.escape(project["tagline"]),
        "video": link_or_placeholder(project["video_url"], "▶ צפו בסרטון ההדגמה", "כאן יופיע סרטון ההדגמה"),
        "steps": li(project["steps"]),
        "workflow": html.escape(project["workflow"]),
        "ai_disclosure": html.escape(project["ai_disclosure"]),
        "demo": link_or_placeholder(project["demo_url"], "לדמו החי", "הדמו יתווסף בהמשך"),
        "limits": li(project["limits"]),
    }
    page = TEMPLATE
    for key, value in values.items():
        page = page.replace("{{" + key + "}}", value)
    return page
````

## שלב 3 - בודקים עמוד HTML

כדי לבדוק HTML לא מחפשים טקסט "בעיניים", וגם לא משתמשים בחיפוש פשוט: **מנתחים אותו**. הספרייה `html.parser` (מובנית בפייתון) קוראת את העמוד תג אחר תג. המחלקה `PageFacts` כבר כתובה: היא קוראת עמוד ו**אוספת עובדות** - אל תשנו אותה, רק קראו והריצו את הדוגמה:

</div>

````python
class PageFacts(HTMLParser):
    """Reads an HTML page and collects facts about it. Written for you - do not change."""
    def __init__(self):
        super().__init__()
        self.title = ""        # text inside <title>
        self.h1_texts = []     # the text of every <h1>
        self.images = []       # one dict of attributes per <img>, e.g. {"src": "a.png", "alt": "logo"}
        self.links = []        # one dict of attributes per <a>,   e.g. {"href": "https://..."}
        self.ids = []          # every id="..." on the page
        self._inside = None

    def handle_starttag(self, tag, attrs):
        attrs = dict(attrs)                       # [("href", "x")] -> {"href": "x"}
        if "id" in attrs:
            self.ids.append(attrs["id"])
        if tag == "img":
            self.images.append(attrs)
        if tag == "a":
            self.links.append(attrs)
        if tag == "h1":
            self.h1_texts.append("")
        if tag in ("title", "h1"):
            self._inside = tag

    def handle_endtag(self, tag):
        if tag in ("title", "h1"):
            self._inside = None

    def handle_data(self, data):
        if self._inside == "title":
            self.title += data
        elif self._inside == "h1":
            self.h1_texts[-1] += data

sample = '<title>דוגמה</title><h1>כותרת</h1><img src="a.png"><a href="">ריק</a><p id="x">t</p><!-- <a href="#"> -->'
facts = PageFacts()
facts.feed(sample)
print("title:", facts.title, "| h1:", facts.h1_texts, "| images:", facts.images, "| links:", facts.links, "| ids:", facts.ids)
````

<div dir="rtl">

שימו לב: הקישור שבתוך ההערה `<!-- ... -->` **לא נספר** - הפרסר יודע שזו הערה. (חיפוש טקסט פשוט היה טועה.) ובתמונה `<img src="a.png">` אין `alt`: המילון שלה הוא `{"src": "a.png"}`.

עכשיו ממשו `validate_html(page)`: מחזירה **רשימה של קודי בעיות** (כל קוד פעם אחת). הכללים:

| קוד | הבעיה | איך בודקים בעובדות |
|---|---|---|
| `no_title` | אין `<title>` או שהוא ריק | `facts.title.strip()` |
| `no_h1` | אין `<h1>` עם טקסט | `facts.h1_texts` |
| `img_no_alt` | תמונה בלי `alt`, או עם `alt` ריק (נגישות: מי שלא רואה את התמונה) | `img.get("alt")` על כל מילון ב-`facts.images` |
| `empty_link` | קישור `<a>` בלי `href`, או `href` ריק, או רק `#` | `a.get("href")` על כל מילון ב-`facts.links` |
| `no_ai_disclosure` | אין אלמנט עם `id="ai-disclosure"` (הצהרת AI, מהצ'קליסט) | `"ai-disclosure" in facts.ids` |

**כלים:** `facts = PageFacts()` ואז `facts.feed(page)`. `d.get("alt")` מחזיר `None` אם המפתח חסר; `(d.get("alt") or "").strip()` מטפל גם ב-`None` וגם ברווחים. `any(... for x in list)`.

</div>

````python
def validate_html(page):
    """page: a string of HTML. Returns a list of problem codes (each at most once; empty list = no problem):
    "no_title", "no_h1", "img_no_alt", "empty_link", "no_ai_disclosure"  (see the table above).
    Use PageFacts: facts = PageFacts(); facts.feed(page).
    """
    facts = PageFacts()
    facts.feed(page)
    problems = []
    if not facts.title.strip():
        problems.append("no_title")
    if not any(t.strip() for t in facts.h1_texts):
        problems.append("no_h1")
    if any(not (img.get("alt") or "").strip() for img in facts.images):
        problems.append("img_no_alt")
    if any((a.get("href") or "").strip() in ("", "#") for a in facts.links):
        problems.append("empty_link")
    if "ai-disclosure" not in facts.ids:
        problems.append("no_ai_disclosure")
    return problems
````

````python
# בדיקה - חלק ב', שלב 3
OK_PAGE = '<html><head><title>x</title></head><body><h1>t</h1><img src="a.png" alt="לוגו"><a href="https://a.test">k</a><p id="ai-disclosure">ai</p></body></html>'

def check_b3():
    assert validate_html(OK_PAGE) == [], "validate_html: על עמוד תקין צריך להחזיר רשימה ריקה"
    assert validate_html(OK_PAGE.replace("<title>x</title>", "")) == ["no_title"], "validate_html: חסר <title>"
    assert validate_html(OK_PAGE.replace("<title>x</title>", "<title> </title>")) == ["no_title"], "validate_html: <title> שיש בו רק רווח נחשב ריק"
    assert validate_html(OK_PAGE.replace("<h1>t</h1>", "<h1> </h1>")) == ["no_h1"], "validate_html: <h1> ריק"
    assert validate_html(OK_PAGE.replace("<h1>t</h1>", "")) == ["no_h1"], "validate_html: אין <h1>"
    assert validate_html(OK_PAGE.replace(' alt="לוגו"', "")) == ["img_no_alt"], "validate_html: תמונה בלי alt"
    assert validate_html(OK_PAGE.replace('alt="לוגו"', 'alt=""')) == ["img_no_alt"], "validate_html: alt ריק נחשב חסר"
    assert validate_html(OK_PAGE.replace('href="https://a.test"', 'href=""')) == ["empty_link"], "validate_html: href ריק"
    assert validate_html(OK_PAGE.replace('href="https://a.test"', 'href="#"')) == ["empty_link"], "validate_html: href=\"#\" נחשב ריק"
    assert validate_html(OK_PAGE.replace('<a href="https://a.test">', "<a>")) == ["empty_link"], "validate_html: <a> בלי href בכלל"
    assert validate_html(OK_PAGE.replace(' id="ai-disclosure"', "")) == ["no_ai_disclosure"], "validate_html: חסר id=\"ai-disclosure\""
    assert validate_html(OK_PAGE.replace("<body>", '<body><!-- <img src="x.png"> <a href=""> -->')) == [], \
        "validate_html: תגים בתוך הערה <!-- --> לא נספרים (השתמשתם ב-PageFacts?)"
    bad = validate_html('<p>hi</p><img src="a.png"><a href="">x</a>')
    assert sorted(bad) == ["empty_link", "img_no_alt", "no_ai_disclosure", "no_h1", "no_title"], "validate_html: צריך לדווח על כל הבעיות, כל אחת פעם אחת"
    return True

check_b3() and print("✓ שלב 3: הבודק תופס את חמש הבעיות")
````

## כללי ההכרעה על מקרים

עכשיו נהפוך את החשיבה לפונקציה `decide(case)` שמקבלת תיאור מובנה של מקרה ומחזירה זוג: **(סיכון, פעולה)**. כל מקרה הוא מילון עם שדות:

| שדה | משמעות |
|---|---|
| `data_sent` | מה הוכנס לכלי AI חיצוני: `"none"`, `"synthetic"`, `"personal"` (מידע אישי), `"secret"` (מפתח/סיסמה/קוד סודי) |
| `tool_approved` | האם הכלי **מאושר** בארגון (`True`/`False`) |
| `copied_license` | רישיון הקוד שהפלט דומה לו: `"none"` (אין), `"MIT"`, `"Apache-2.0"`, `"BSD"`, `"GPL"`, `"AGPL"`, `"LGPL"`, `"missing"` (מאגר בלי LICENSE) |
| `chatbot_promise` | האם צ'אטבוט התחייב להבטחה שאינה במדיניות |
| `customer_facing`, `ai_disclosed` | האם הכלי מדבר עם לקוחות, והאם הם יודעים שזה AI |
| `decides_about_people`, `human_approves` | האם AI מחליט על אנשים (גיוס, אשראי), והאם אדם מאשר |

**הכללים, לפי הסדר** (הראשון שמתקיים מכריע; הסיבה לסדר: נזק שאי אפשר להחזיר קודם):

1. **דליפה:** מפתח/סוד שהוכנס לכלי (גם מאושר), או מידע אישי שהוכנס לכלי **לא** מאושר ← `("privacy", "report_and_assess")`. (מידע אישי בכלי מאושר תלוי בבסיס חוקי ובהסכם עיבוד, לבדיקת עורך דין: מחוץ לכלל הפשוט שלנו.)
2. **רישיון:** לפי `copied_license`: `MIT`/`Apache-2.0`/`BSD` ← `("copyright", "keep_notice")`; `GPL`/`AGPL`/`LGPL` ← `("copyright", "check_source_and_replace")`; `missing` ← `("copyright", "do_not_use_without_permission")`.
3. **אחריות:** `chatbot_promise` ← `("liability", "check_with_lawyer")`.
4. **גילוי:** פונה ללקוחות **ולא** הודיע שזה AI ← `("transparency", "add_disclosure")`.
5. **פיקוח אנושי:** AI מחליט על אנשים **בלי** אישור אדם ← `("high_risk", "require_human_approval")`.
6. אחרת ← `("none", "proceed")`.

נחלק לשלוש פונקציות קטנות. **שלב 2 - `privacy_leak(case)`** ← `True`/`False` לפי כלל 1. **כלים:** `case["data_sent"] == "secret"`, `and`, `or`, `not`.

</div>

````python
def case(text, data_sent="none", tool_approved=True, copied_license="none", chatbot_promise=False,
         customer_facing=False, ai_disclosed=True, decides_about_people=False, human_approves=True):
    """Builds a case dict. Written for you. Fields that are not given get a harmless default."""
    return dict(text=text, data_sent=data_sent, tool_approved=tool_approved, copied_license=copied_license,
                chatbot_promise=chatbot_promise, customer_facing=customer_facing, ai_disclosed=ai_disclosed,
                decides_about_people=decides_about_people, human_approves=human_approves)

def privacy_leak(case):
    """True if data leaked: a "secret" went into a tool (approved or not),
    or "personal" data went into a tool that is NOT approved. Otherwise False.
    """
    return case["data_sent"] == "secret" or (case["data_sent"] == "personal" and not case["tool_approved"])
````

````python
# בדיקה - חלק ג', שלב 2
def check_c2():
    assert privacy_leak(case("", data_sent="personal", tool_approved=False)) is True, "privacy_leak: מידע אישי בכלי לא מאושר הוא דליפה"
    assert privacy_leak(case("", data_sent="secret", tool_approved=True)) is True, "privacy_leak: מפתח/סוד בכלי הוא דליפה גם אם הכלי מאושר"
    assert privacy_leak(case("", data_sent="secret", tool_approved=False)) is True, "privacy_leak: סוד בכלי לא מאושר - דליפה"
    assert privacy_leak(case("", data_sent="synthetic", tool_approved=False)) is False, "privacy_leak: נתונים סינתטיים אינם דליפה"
    assert privacy_leak(case("", data_sent="none")) is False, "privacy_leak: לא הוכנס כלום - אין דליפה"
    assert privacy_leak(case("", data_sent="personal", tool_approved=True)) is False, "privacy_leak: מידע אישי בכלי מאושר - מחוץ לכלל הפשוט (לבדיקת עורך דין), לא דליפה לפי הכלל שלנו"
    return True

check_c2() and print("✓ שלב 2: privacy_leak")
````

<div dir="rtl">

**שלב 3 - `license_action(lic)`** לפי כלל 2. מחזירה את **טקסט הפעולה**, או `None` אם אין קוד מועתק (`"none"`). **כלים:** `lic in ("MIT", "Apache-2.0", "BSD")` בודק אם הערך אחד מכמה אפשרויות.

</div>

````python
def license_action(lic):
    """"MIT"/"Apache-2.0"/"BSD" -> "keep_notice"; "GPL"/"AGPL"/"LGPL" -> "check_source_and_replace";
    "missing" -> "do_not_use_without_permission"; anything else ("none") -> None.
    """
    if lic in ("MIT", "Apache-2.0", "BSD"):
        return "keep_notice"
    if lic in ("GPL", "AGPL", "LGPL"):
        return "check_source_and_replace"
    if lic == "missing":
        return "do_not_use_without_permission"
    return None
````

````python
# בדיקה - חלק ג', שלב 3
def check_c3():
    assert [license_action(x) for x in ("MIT", "Apache-2.0", "BSD")] == ["keep_notice"] * 3, "license_action: רישיון מתירני - לשמור את הודעת הזכויות (keep_notice)"
    assert [license_action(x) for x in ("GPL", "AGPL", "LGPL")] == ["check_source_and_replace"] * 3, "license_action: קופילפט - לבדוק מקור ולהחליף (check_source_and_replace)"
    assert license_action("missing") == "do_not_use_without_permission", "license_action: בלי LICENSE ברירת המחדל היא כל הזכויות שמורות"
    assert license_action("none") is None, "license_action: אין קוד מועתק - None"
    return True

check_c3() and print("✓ שלב 3: license_action")
````

<div dir="rtl">

**שלב 4 - הפונקציה הראשית `decide(case)`**: מחזירה `(risk, action)` לפי ששת הכללים **בסדר שלמעלה**. שלד: `if privacy_leak(case): return ("privacy", "report_and_assess")`, ואחר כך `action = license_action(case["copied_license"])` - אם הוא לא `None`, מחזירים `("copyright", action)`, וכן הלאה. את שלושת הכללים האחרונים כתבו בעצמכם לפי הרשימה.

</div>

````python
def decide(case):
    """Applies the six rules in order (first match wins). Returns a tuple (risk, action):
    ("privacy", "report_and_assess"), ("copyright", <license_action>), ("liability", "check_with_lawyer"),
    ("transparency", "add_disclosure"), ("high_risk", "require_human_approval"), or ("none", "proceed").
    """
    if privacy_leak(case):
        return ("privacy", "report_and_assess")
    action = license_action(case["copied_license"])
    if action:
        return ("copyright", action)
    if case["chatbot_promise"]:
        return ("liability", "check_with_lawyer")
    if case["customer_facing"] and not case["ai_disclosed"]:
        return ("transparency", "add_disclosure")
    if case["decides_about_people"] and not case["human_approves"]:
        return ("high_risk", "require_human_approval")
    return ("none", "proceed")
````

<div dir="rtl">

כשקבוצה כותבת מדיניות (בכיתה, בעזרת AI), קל לפספס סעיף, או לכתוב סעיף שהוא רק משפט כללי. נכתוב **בודק מדיניות**: `check_policy(text)` שמוודא שיש את **כל שמונת הסעיפים** ושבכל סעיף יש תוכן, ו-`policy_extras(text)` שבודקת שני דברים שהשיעור הדגיש: **סיווג מידע עם שלושת הצבעים**, ו**סימון נושא משפטי כ"לבדיקת עורך דין"**.

הפונקציה `section_bodies(text)` כבר כתובה: היא מוצאת בטקסט **שורה שמתחילה בשם של סעיף** (גם אם יש לפניה מספר, `#` או `**`) ומחזירה מילון: שם סעיף ← **הטקסט שמתחת לכותרת שלו**, או `None` אם אין כזה סעיף. רשימת הסעיפים (`SECTIONS`, בקיצור) כתובה גם היא. הריצו וראו מה היא מחזירה (שימו לב: גם שארית השורה של הכותרת, כמו `והיקף:**`, נכללת בגוף - זה בסדר):

</div>

````python
SECTIONS = {  # שם קצר של הסעיף -> התבנית שמחפשים בתחילת שורה
    "מטרה": "מטרה",
    "כלים מאושרים": "כלים מאושרים",
    "סיווג מידע": "סיווג מידע",
    "סקירה אנושית": "סקירה אנושית",
    "אבטחה": "אבטחה",
    "קניין רוחני": "קני{1,2}ן רוחני",
    "שקיפות": "שקיפות",
    "אירועים": "אירועים",
}

def section_bodies(text):
    """For each name in SECTIONS: the text under its heading (until the next heading), or None if missing.
    A heading is a line that starts with the name (after an optional number, # or **). Written for you."""
    found = {}
    for name, pattern in SECTIONS.items():
        m = re.search(r"^[^\w\n]*\d*[.)]?[^\w\n]*" + pattern, text, re.MULTILINE)
        if m:
            found[name] = (m.start(), m.end())
    starts = sorted(start for start, end in found.values())
    bodies = {}
    for name in SECTIONS:
        if name not in found:
            bodies[name] = None
        else:
            start, end = found[name]
            later = [s for s in starts if s > start]
            bodies[name] = text[end:(later[0] if later else len(text))]
    return bodies

demo = "1. **מטרה והיקף:** על מי חלה המדיניות\n2. **כלים מאושרים:** רק כלים מהרשימה של הארגון"
for name, body in section_bodies(demo).items():
    print(f"{name}: {body!r}")
````

## שלב 1 - הבודק

ממשו `check_policy(text)`: מחזירה **רשימה של בעיות** (טקסטים), ריקה אם הכול תקין. לכל אחד משמונת הסעיפים (בלולאה על `SECTIONS`):

- אין סעיף כזה (`bodies[name] is None`) ← מוסיפים `"missing: " + name`.
- יש, אבל **דל**: פחות מ-8 מילים בגוף שלו (`len(body.split()) < 8`) ← מוסיפים `"thin: " + name`. (מדיניות שלא אומרת כלום אינה מדיניות.)

ואז `policy_extras(text)` - מחזירה רשימת בעיות נוספת:

- בסעיף "סיווג מידע" (אם קיים) חסר אחד מהצבעים **אדום, צהוב, ירוק** ← `"no_colors"`.
- בכל הטקסט אין אף אחת מהמילים `"עורך דין"`, `"יועץ משפטי"`, `"lawyer"` ← `"no_lawyer_note"`. (בהוראה ל-AI ביקשנו לסמן נושא משפטי כ"דורש בדיקת עורך דין": זו התנהגות רצויה, לא חולשה.)

**כלים:** `bodies = section_bodies(text)`; `for name in SECTIONS:`; `all(c in body for c in ("אדום", "צהוב", "ירוק"))`; `any(w in text for w in (...))`. את הגוף של סעיף יכול להיות `None`, ולכן בדקו `if body and not all(...)`.

</div>

````python
def check_policy(text):
    """Returns a list of problems (empty list = fine). For each name in SECTIONS:
    section missing -> "missing: <name>";  section body shorter than 8 words -> "thin: <name>".
    Use bodies = section_bodies(text).
    """
    bodies = section_bodies(text)
    problems = []
    for name in SECTIONS:
        if bodies[name] is None:
            problems.append("missing: " + name)
        elif len(bodies[name].split()) < 8:
            problems.append("thin: " + name)
    return problems


def policy_extras(text):
    """Returns a list of extra problems:
    "no_colors" - the "סיווג מידע" section exists but lacks one of: אדום, צהוב, ירוק
    "no_lawyer_note" - none of "עורך דין", "יועץ משפטי", "lawyer" appears anywhere in the text
    """
    problems = []
    body = section_bodies(text)["סיווג מידע"]
    if body and not all(color in body for color in ("אדום", "צהוב", "ירוק")):
        problems.append("no_colors")
    if not any(w in text for w in ("עורך דין", "יועץ משפטי", "lawyer")):
        problems.append("no_lawyer_note")
    return problems
````

````python
# בדיקה - חלק ד', שלב 1
FULL_TITLES = ["מטרה והיקף", "כלים מאושרים", "סיווג מידע", "סקירה אנושית", "אבטחה",
               "קניין רוחני ורישיונות", "שקיפות ותיעוד", "אירועים והכשרה"]
FILLER = "כלל קונקרטי אחד שמסביר מי עושה מה ומתי ואיך בודקים זאת בפועל בארגון"
COLORS = "אדום לקוחות וסודות צהוב באישור ירוק נתונים סינתטיים וקוד ציבורי בלבד"

def make_policy(skip=(), style="plain", thin=(), colors=True, lawyer=True):
    lines = []
    for i, title in enumerate(FULL_TITLES, 1):
        if title in skip:
            continue
        body = "קצר" if title in thin else (COLORS if (title == "סיווג מידע" and colors) else FILLER)
        if title == "סיווג מידע" and not colors:
            body = FILLER
        head = {"plain": f"{i}. {title}:", "md": f"## {i}. **{title}**", "bold": f"**{title}:**"}[style]
        lines.append(f"{head} {body}" if style != "md" else f"{head}\n{body}")
    if lawyer:
        lines.append("הערה: כל נושא משפטי דורש בדיקת עורך דין.")
    return "\n".join(lines)

def check_d1():
    for style in ("plain", "md", "bold"):
        assert check_policy(make_policy(style=style)) == [], f"check_policy: מדיניות מלאה (סגנון כותרת {style}) צריכה להחזיר רשימה ריקה"
    assert check_policy(make_policy(skip=("אבטחה", "אירועים והכשרה"))) == ["missing: אבטחה", "missing: אירועים"], \
        "check_policy: שני סעיפים חסרים - missing: <שם קצר מ-SECTIONS>, לפי סדר SECTIONS"
    assert check_policy(make_policy(thin=("סקירה אנושית",))) == ["thin: סקירה אנושית"], "check_policy: סעיף עם פחות מ-8 מילים הוא thin"
    assert check_policy(make_policy().replace("קניין", "קנין")) == [], "check_policy: כתיב ״קנין״ בלי יוד כפולה צריך להתקבל (זה כבר מטופל ב-section_bodies)"
    assert policy_extras(make_policy()) == [], "policy_extras: מדיניות עם צבעים והערת עורך דין - רשימה ריקה"
    assert policy_extras(make_policy(colors=False)) == ["no_colors"], "policy_extras: בסעיף סיווג מידע חסרים הצבעים"
    assert policy_extras(make_policy(lawyer=False)) == ["no_lawyer_note"], "policy_extras: אין התייחסות לעורך דין"
    assert policy_extras(make_policy(skip=("סיווג מידע",))) == [], "policy_extras: אם אין סעיף סיווג - זו בעיה של check_policy, לא של no_colors"
    return True

check_d1() and print("✓ חלק ד' שלב 1: הבודק תופס סעיפים חסרים, דלים, וצבעים והערה משפטית")
````
