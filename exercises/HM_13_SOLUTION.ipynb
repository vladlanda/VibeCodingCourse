## שלב 1 - `weighted_score`

ממשו `weighted_score(rubric, scores)`: `rubric` הוא מילון `{ממד: משקל}`, `scores` מילון `{ממד: ציון 0-5}`. הפונקציה מחזירה מספר בין 0 ל-100.

**הרעיון (נרמול):** מחשבים כמה נקודות קיבלה הקבוצה (כל ציון כפול המשקל שלו, מסכמים), ומחלקים בכמה נקודות היא **יכולה** לקבל לכל היותר (המשקלים כפול הציון המקסימלי). את התוצאה הופכים לאחוזים.

**כלים:** `sum(...)` עם ביטוי בתוך הסוגריים, למשל `sum(x * 2 for x in [1, 2, 3])` נותן 12. `rubric.values()` מחזיר את המשקלים בלבד. `rubric[c]` ו-`scores[c]` נותנים משקל וציון של ממד `c`.

</div>

````python
def weighted_score(rubric, scores):
    """rubric: {criterion: weight}, scores: {criterion: score 0..MAX_SCORE}.
    Returns the normalized weighted score: a number between 0 and 100.
    e.g. weighted_score({"a": 1, "b": 3}, {"a": 5, "b": 0}) -> 25.0
    """
    earned = sum(rubric[c] * scores[c] for c in rubric)
    possible = sum(rubric.values()) * MAX_SCORE
    return earned / possible * 100
````

````python
# בדיקה - חלק א', שלב 1
def check_a1():
    assert abs(weighted_score({"a": 1, "b": 3}, {"a": 5, "b": 0}) - 25.0) < 0.01, "weighted_score: הדוגמה מה-docstring - (1*5 + 3*0) מתוך (4*5) = 25%"
    assert abs(weighted_score({"x": 10, "y": 10}, {"x": 5, "y": 5}) - 100.0) < 0.01, "weighted_score: ציון מקסימלי בכל הממדים צריך להיות 100, גם כשסכום המשקלים הוא לא 100 (נרמול!)"
    assert abs(weighted_score({"x": 10, "y": 10}, {"x": 0, "y": 0}) - 0.0) < 0.01, "weighted_score: אפס בכל הממדים צריך להיות 0"
    assert abs(weighted_score({"x": 1, "y": 1}, {"x": 5, "y": 0}) - 50.0) < 0.01, "weighted_score: חצי מהנקודות האפשריות = 50"
    full = {c: 5 for c in RUBRIC}
    assert abs(weighted_score(RUBRIC, full) - 100.0) < 0.01, "weighted_score: על RUBRIC עם 5 בכל ממד צריך לקבל 100"
    heavy = dict(full, **{"סוכן שעובד": 0})
    assert abs(weighted_score(RUBRIC, heavy) - 75.0) < 0.01, "weighted_score: אפס בממד שמשקלו 25 מתוך 100 מוריד בדיוק 25 נקודות"
    return True

check_a1() and print("✓ חלק א' שלב 1: הציון המשוקלל מחושב ומנורמל")
````

## שלב 2 - לאמת את הקלט: `score_problems`

מחוון מחזיר ציון גם כשהקלט שבור: ציון שנשכח, ציון 7 מתוך 5, או שם ממד עם שגיאת כתיב. **"אמתו"** חל גם על הכלים שלנו. ממשו `score_problems(rubric, scores)` שמחזירה **רשימה של בעיות** (טקסטים; רשימה ריקה = הקלט תקין), **בעיה אחת לכל ממד בעייתי**:

1. ממד שיש ב-`rubric` ואין ב-`scores` ← "חסר ציון".
2. ציון שאינו בין 0 ל-`MAX_SCORE` (כולל) ← "ציון מחוץ לטווח".
3. ממד שיש ב-`scores` ואין ב-`rubric` ← "ממד לא מוכר".

**כלים:** `for c in rubric:` עובר על שמות הממדים. `c not in scores` בודק אם מפתח חסר במילון. `0 <= x <= 5` בודק טווח. `problems.append("...")` מוסיף לרשימה.

</div>

````python
def score_problems(rubric, scores):
    """Returns a list of problem texts (empty list = the scores are valid), one per bad criterion:
    a criterion of the rubric with no score; a score outside 0..MAX_SCORE; a score for an unknown criterion.
    """
    problems = []
    for c in rubric:
        if c not in scores:
            problems.append(f"חסר ציון: {c}")
        elif not 0 <= scores[c] <= MAX_SCORE:
            problems.append(f"ציון מחוץ לטווח ב-{c}: {scores[c]}")
    for c in scores:
        if c not in rubric:
            problems.append(f"ממד לא מוכר: {c}")
    return problems
````

````python
# בדיקה - חלק א', שלב 2
def check_a2():
    ok = {c: 3 for c in RUBRIC}
    assert score_problems(RUBRIC, ok) == [], "score_problems: על קלט תקין צריך להחזיר רשימה ריקה"
    missing = {c: 3 for c in list(RUBRIC)[:-1]}
    assert len(score_problems(RUBRIC, missing)) == 1, "score_problems: ממד אחד חסר = בעיה אחת"
    two_missing = {c: 3 for c in list(RUBRIC)[:-2]}
    assert len(score_problems(RUBRIC, two_missing)) == 2, "score_problems: שני ממדים חסרים = שתי בעיות"
    high = dict(ok, **{"עמוד נחיתה": 7})
    assert len(score_problems(RUBRIC, high)) == 1, "score_problems: ציון 7 (מעל 5) הוא בעיה אחת"
    low = dict(ok, **{"עמוד נחיתה": -1})
    assert len(score_problems(RUBRIC, low)) == 1, "score_problems: ציון שלילי הוא בעיה"
    edges = dict(ok, **{"עמוד נחיתה": 0, "סוכן שעובד": 5})
    assert score_problems(RUBRIC, edges) == [], "score_problems: 0 ו-5 הם ציונים תקינים (הקצוות כלולים)"
    extra = dict(ok, **{"עמוד נחיתה ": 4})
    assert len(score_problems(RUBRIC, extra)) == 1, "score_problems: ממד עם שם שגוי (רווח מיותר) הוא ממד לא מוכר"
    return True

check_a2() and print("✓ חלק א' שלב 2: הקלט נבדק לפני שמחשבים ציון")
````

## שלב 1 - לפרק README לסעיפים: `split_sections`

ממשו `split_sections(text)` שמחזירה **מילון**: המפתח הוא הטקסט של כותרת (בלי סימני `#`, בלי רווחים בקצוות, באותיות קטנות), והערך הוא כל הטקסט שמתחתיה (עד הכותרת הבאה). שורות שלפני הכותרת הראשונה מתעלמים מהן.

**איך:** עוברים על `text.splitlines()`. שורה שמתחילה ב-`"#"` (`line.startswith("#")`) פותחת סעיף חדש: `line.lstrip("#").strip().lower()` נותן את שם הכותרת. כל שורה אחרת מצטרפת לטקסט של הסעיף **הנוכחי** (שמרו את שם הסעיף הנוכחי במשתנה, ואפשר להתחיל ב-`None`).

דוגמה: הטקסט `"# A\nhello\n## B c\none two"` נותן `{"a": "hello", "b c": "one two"}` (פרט לרווחים).

</div>

````python
def split_sections(text):
    """Splits a markdown text into sections.
    Returns {heading_text_lowercase: body_text}. Lines before the first heading are ignored.
    e.g. split_sections("# A\nhello\n## B c\none two") -> {"a": "hello", "b c": "one two"}
    (the exact spaces/newlines inside a body do not matter)
    """
    sections = {}
    current = None
    for line in text.splitlines():
        if line.startswith("#"):
            current = line.lstrip("#").strip().lower()
            sections[current] = ""
        elif current is not None:
            sections[current] += line + " "
    return sections
````

````python
# בדיקה - חלק ב', שלב 1
def check_b1():
    s = split_sections("ignored line\n# A\nhello\n## B C \none two\nthree")
    assert set(s) == {"a", "b c"}, "split_sections: המפתחות צריכים להיות טקסט הכותרת באותיות קטנות, בלי # ובלי רווחים בקצוות (ושורות לפני הכותרת הראשונה מתעלמים מהן)"
    assert s["a"].split() == ["hello"], "split_sections: הטקסט של הסעיף a הוא hello"
    assert s["b c"].split() == ["one", "two", "three"], "split_sections: הטקסט של הסעיף כולל את כל השורות עד הכותרת הבאה"
    he = split_sections("## הבעיה\nמנהל המלאי מגלה חוסרים מאוחר")
    assert list(he) == ["הבעיה"], "split_sections: כותרת בעברית צריכה לעבוד בדיוק כמו באנגלית"
    assert split_sections("רק טקסט בלי כותרות") == {}, "split_sections: בלי כותרות מחזירים מילון ריק"
    return True

check_b1() and print("✓ חלק ב' שלב 1: README מתפרק לסעיפים")
````

## שלב 2 - `check_readme`

ממשו `check_readme(text)` שמחזירה **רשימה של שמות הסעיפים החסרים** (המפתחות של `REQUIRED`, **בסדר שבו הם מופיעים ב-`REQUIRED`**). רשימה ריקה = ה-README מלא.

סעיף נחשב **קיים** אם יש כותרת שמכילה **אחת** ממילות המפתח שלו, **וגם** מתחתיה לפחות `MIN_WORDS` מילים.

**איך:**
1. `sections = split_sections(strip_code_blocks(text))` (קודם מסירים בלוקי קוד, כדי שהערות בתוכם לא ייחשבו כותרות).
2. לכל סעיף ב-`REQUIRED.items()`: עוברים על `sections.items()` (כותרת וטקסט). הכותרת מתאימה אם `any(k in heading for k in keywords)`. מספר המילים הוא `len(body.split())`.
3. סעיף שלא נמצא ← `missing.append(name)`.

</div>

````python
def check_readme(text):
    """Returns the list of REQUIRED section names that are missing from the README text
    (same order as in REQUIRED). A section exists if some heading contains one of its keywords
    AND has at least MIN_WORDS words under it. Code blocks are ignored.
    """
    sections = split_sections(strip_code_blocks(text))
    missing = []
    for name, keywords in REQUIRED.items():
        found = False
        for heading, body in sections.items():
            if any(k in heading for k in keywords) and len(body.split()) >= MIN_WORDS:
                found = True
        if not found:
            missing.append(name)
    return missing
````

## שלב 1 - לקרוא את היומן: `parse_log`

ממשו `parse_log(text)` שמקבלת את טקסט הקובץ ומחזירה **רשימה של מילונים**, מילון אחד לכל שורת נתונים, עם המפתחות `tool, prompt, got, verified, decision` (הרשימה `KEYS` מוגדרת למטה). טקסט כל תא **בלי רווחים בקצוות**.

**איך:**
1. שורות הטבלה הן השורות שמתחילות ב-`|` (אחרי `line.strip()`). שימו אותן ברשימה. שתי הראשונות הן **כותרת והפרדה** (`|---|`), וקובעים מהן את `[2:]`.
2. כל שורה מפרקים לתאים: `line.strip().strip("|").split("|")`, ולכל תא עושים `.strip()`.
3. `dict(zip(KEYS, cells))` בונה מילון מרשימת מפתחות ורשימת תאים.

שימו לב: תא ריק (איך אימתתי) חייב להישאר **מחרוזת ריקה** ולא להיעלם.

</div>

````python
KEYS = ["tool", "prompt", "got", "verified", "decision"]
EMPTY_MARKS = {"", "-", "—", "אין", "none", "n/a"}   # a "verified" cell with one of these = NOT verified


def classify_decision(text):
    """Returns "adopted", "rejected", "fixed" or "other". Already written for you."""
    t = text.strip().lower()
    if t.startswith(("אימצ", "adopt")):
        return "adopted"
    if t.startswith(("דחי", "reject")):
        return "rejected"
    if t.startswith(("תיקנ", "fix", "modif")):
        return "fixed"
    return "other"


print("✓ הוגדרו KEYS, EMPTY_MARKS, classify_decision")
````

````python
def parse_log(text):
    """Reads an AI_LOG.md text. Returns a list of dicts (one per data row) with the keys in KEYS.
    Only table lines (starting with "|") count; the first two are the header and the separator.
    Every cell is stripped; an empty cell stays "".
    """
    table = [line.strip() for line in text.splitlines() if line.strip().startswith("|")]
    rows = []
    for line in table[2:]:
        cells = [c.strip() for c in line.strip("|").split("|")]
        rows.append(dict(zip(KEYS, cells)))
    return rows
````

## שלב 2 - לספור: `ai_use_summary`

- `adopted`, `rejected`, `fixed` - כמה פעמים כל החלטה (השתמשו ב-`classify_decision(row["decision"])`, שמחזירה `"adopted"`, `"rejected"`, `"fixed"` או `"other"`)
- `unverified` - כמה שורות **בלי אימות**: `row["verified"].strip().lower() in EMPTY_MARKS`

**איך:** מתחילים ממילון עם אפסים, עוברים בלולאה על השורות, ו-`summary[kind] += 1` מגדיל מונה. שורה עם החלטה `"other"` נספרת רק ב-`total`.

</div>

````python
def ai_use_summary(rows):
    """rows: the list from parse_log. Returns
    {"total": n, "adopted": n, "rejected": n, "fixed": n, "unverified": n}
    where unverified = rows whose "verified" cell is empty (see EMPTY_MARKS).
    """
    summary = {"total": len(rows), "adopted": 0, "rejected": 0, "fixed": 0, "unverified": 0}
    for row in rows:
        kind = classify_decision(row["decision"])
        if kind in summary:
            summary[kind] += 1
        if row["verified"].strip().lower() in EMPTY_MARKS:
            summary["unverified"] += 1
    return summary
````

# חלק ד' - משוב עמיתים מובנה (2 דק')

## הרקע

בשקף "טופס משוב לעמיתים": לכל קבוצה ממלאים שלוש נקודות - **מה עבד הכי טוב** (נקודה אחת, ספציפית), **איזו שאלה נשארה לי**, **שיפור אחד שהייתי מציע**. הכלל: **ספציפי, מנומס, ומתייחס לעבודה ולא לאנשים**. נכתוב פונקציה שמרכיבה את הטופס ו**מסרבת** למשוב ריק או כללי מדי ("יפה מאוד").

## שלב 1 - `format_feedback`

ממשו `format_feedback(strengths, questions, suggestions)` שמחזירה טקסט של **שלוש שורות** בדיוק (הפרדה ב-`\n`):

```
1. מה עבד הכי טוב? <strengths>
2. איזו שאלה נשארה לי? <questions>
3. שיפור אחד שהייתי מציע: <suggestions>
```

**ספציפי = יותר ממילה אחת:** אם אחד משלושת הטקסטים כולל **פחות מ-3 מילים** (`len(text.split())`), הפונקציה זורקת שגיאה: `raise ValueError("הודעה")`.

</div>

````python
def format_feedback(strengths, questions, suggestions):
    """Returns the peer-feedback form as 3 lines joined by "\n":
    "1. מה עבד הכי טוב? <strengths>"
    "2. איזו שאלה נשארה לי? <questions>"
    "3. שיפור אחד שהייתי מציע: <suggestions>"
    Raises ValueError if any of the three texts has fewer than 3 words (feedback must be specific).
    """
    for text in (strengths, questions, suggestions):
        if len(text.split()) < 3:
            raise ValueError("משוב צריך להיות ספציפי: לפחות 3 מילים בכל נקודה")
    return (f"1. מה עבד הכי טוב? {strengths}\n"
            f"2. איזו שאלה נשארה לי? {questions}\n"
            f"3. שיפור אחד שהייתי מציע: {suggestions}")
````

````python
# בדיקה - חלק ד', שלב 1
def check_d1():
    r = format_feedback("הדמו הראה בבירור את ההתראה", "איפה נשמר המפתח", "להוסיף שקף על מגבלות")
    lines = r.splitlines()
    assert len(lines) == 3, "format_feedback: צריכות להיות 3 שורות בדיוק"
    assert lines[0] == "1. מה עבד הכי טוב? הדמו הראה בבירור את ההתראה", "format_feedback: שורה 1 - '1. מה עבד הכי טוב? ' ואחריה הטקסט"
    assert lines[1] == "2. איזו שאלה נשארה לי? איפה נשמר המפתח", "format_feedback: שורה 2 - '2. איזו שאלה נשארה לי? ' ואחריה הטקסט"
    assert lines[2] == "3. שיפור אחד שהייתי מציע: להוסיף שקף על מגבלות", "format_feedback: שורה 3 - '3. שיפור אחד שהייתי מציע: ' ואחריה הטקסט"
    for bad in [("יפה מאוד", "איפה נשמר המפתח", "להוסיף שקף על מגבלות"),
                ("הדמו הראה בבירור", "", "להוסיף שקף על מגבלות"),
                ("הדמו הראה בבירור", "איפה נשמר המפתח", "שיפור")]:
        try:
            format_feedback(*bad)
        except ValueError:
            continue
        raise AssertionError(f"format_feedback: משוב קצר מ-3 מילים ('{min(bad, key=lambda t: len(t.split()))}') צריך לזרוק ValueError")
    return True

check_d1() and print("✓ חלק ד' שלב 1: הטופס מורכב, ומשוב כללי נדחה")
````

ממשו `quiz(answers)`: `answers` הוא מילון `{מספר שאלה: אות}`, למשל `{1: "b", 2: "a"}`. הפונקציה עוברת על **כל** עשר השאלות, ו**לכל שאלה מדפיסה משוב מיידי**:

- תשובה נכונה ← מדפיסה שורה שמתחילה ב-`✓` (למשל `✓ שאלה 3`).
- תשובה שגויה **או חסרה** ← שורה שמתחילה ב-`✗`, ואחריה **ההסבר** של השאלה (`item["explain"]`), כדי שתדעו למה.
- בסוף **מחזירה** (`return`) את מספר התשובות הנכונות.

**כלים:** `for i, item in enumerate(QUIZ, start=1):` נותן מספר שאלה ופריט. `answers.get(i, "")` מחזיר את התשובה, או טקסט ריק אם לא נענתה. `.strip().lower()` מנקה (אפשר גם `B` וגם `b`). `is_correct(i, תשובה)` כבר כתובה.

</div>

````python
def quiz(answers):
    """answers: {question_number: letter}. For EVERY question (1..10) prints immediate feedback:
    a line starting with "✓" for a correct answer, or a line starting with "✗" followed by the
    explanation (item["explain"]) for a wrong or missing one. Returns the number of correct answers.
    """
    score = 0
    for i, item in enumerate(QUIZ, start=1):
        letter = str(answers.get(i, "")).strip().lower()
        if is_correct(i, letter):
            score += 1
            print(f"✓ שאלה {i}")
        else:
            print(f"✗ שאלה {i} (מפגש {item['session']}): {item['explain']}")
    return score
````

````python
# בדיקה - חלק ה', שלב 1
def run_quiet(answers):
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        result = quiz(answers)
    return result, buf.getvalue()

def check_e1():
    right = {i: next(l for l in "abc" if is_correct(i, l)) for i in range(1, 11)}
    wrong = {i: next(l for l in "abc" if not is_correct(i, l)) for i in range(1, 11)}
    score, out = run_quiet(right)
    assert score == 10, f"quiz: כל התשובות נכונות - צריך להחזיר 10, התקבל {score}"
    assert "✗" not in out and out.count("✓") == 10, "quiz: על כל תשובה נכונה מדפיסים ✓ (ובלי ✗ כשהכול נכון) - עשר שורות עם ✓"
    score, out = run_quiet(wrong)
    assert score == 0, f"quiz: כל התשובות שגויות - צריך להחזיר 0, התקבל {score}"
    assert out.count("✗") == 10 and all(item["explain"] in out for item in QUIZ), "quiz: על כל תשובה שגויה מדפיסים ✗ וגם את ההסבר של השאלה (item['explain'])"
    mixed = dict(right)
    mixed[3] = wrong[3]
    del mixed[7]
    score, out = run_quiet(mixed)
    assert score == 8, f"quiz: 8 נכונות (אחת שגויה ואחת חסרה) - התקבל {score}. שאלה שלא נענתה נחשבת שגויה"
    assert QUIZ[2]["explain"] in out and QUIZ[6]["explain"] in out, "quiz: גם לשאלה שלא נענתה צריך להדפיס ✗ והסבר"
    upper = {i: l.upper() for i, l in right.items()}
    assert run_quiet(upper)[0] == 10, "quiz: אות גדולה (B) צריכה להיחשב כמו אות קטנה (b) - השתמשו ב-.lower()"
    return True

check_e1() and print("✓ חלק ה' שלב 1: החידון עובד")
````
