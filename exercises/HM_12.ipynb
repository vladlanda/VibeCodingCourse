## שלב 3 - קישור שבור (מבנית)

בעמוד נחיתה ובקובץ README יש קישורים. ב-Colab **אין לנו רשת** כדי לבדוק אם אתר באמת עונה, אבל אפשר לתפוס קישורים ש**שבורים כבר במבנה** שלהם. ממשו `is_broken_link(link, files, folder="")` שמחזירה `True` אם הקישור שבור. הכללים, **לפי הסדר** (הראשון שמתקיים מכריע):

1. הקישור ריק, או רק `#` (אחרי `link.strip()`) ← **שבור**.
2. הקישור מכיל `example.com`, `localhost`, `127.0.0.1` או `TODO` (באותיות גדולות) ← **שבור**: זה placeholder או כתובת מקומית.
3. הקישור מתחיל ב-`#` (עוגן בתוך העמוד, כמו `#demo`), או ב-`http://`, `https://`, `mailto:` ← **תקין**. (לא בודקים שהאתר עונה: את זה בודקים בדפדפן.)
4. אחרת זה קישור **יחסי** לקובץ: שבור אם הקובץ **לא במאגר**. הנתיב המלא הוא `folder + link` (בלי החלק אחרי `#`, אם יש), ו-`folder` הוא התיקייה של הקובץ שבו נמצא הקישור, למשל `"docs/"`.

**כלים:** `link.strip()`; `"x" in link`; `link.startswith(("#", "http://"))` - אפשר לתת **סוגריים עם כמה אפשרויות** (tuple); `link.split("#")[0]` לוקח את מה שלפני `#`; `path not in files` בודק אם נתיב חסר במילון.

הפונקציות `find_links`, `folder_of` ו-`all_links` כבר כתובות: הן מוצאות את כל הקישורים בקבצי `.md` ו-`.html` (קישור ב-HTML: `href="..."`, ב-Markdown: `[טקסט](קישור)`).

</div>

````python
# כתוב בשבילכם - הריצו וקראו
LINK_RE = re.compile(r"""href=["']([^"']*)["']|\]\(([^)]*)\)""")

def find_links(text):
    """All link targets in a text: href="..." (HTML) and [text](...) (Markdown)."""
    return [a or b for a, b in LINK_RE.findall(text)]

def folder_of(name):
    """'docs/index.html' -> 'docs/' ;  'README.md' -> ''"""
    return name.rsplit("/", 1)[0] + "/" if "/" in name else ""

def all_links(files):
    """Every link in every .md / .html file: a list of (link, folder of the file that contains it)."""
    pairs = []
    for name, text in files.items():
        if name.endswith((".md", ".html")):
            pairs += [(link, folder_of(name)) for link in find_links(text)]
    return pairs

print(all_links(store_good))
````

- אם `url` מתחיל ב-`http://` או `https://` ← מחזירה את הטקסט `<a class="button" href="URL">TEXT</a>` (ה-URL והטקסט **אחרי `html.escape`**).

## שלב 3 - בודקים עמוד HTML

כדי לבדוק HTML לא מחפשים טקסט "בעיניים", וגם לא משתמשים בחיפוש פשוט: **מנתחים אותו**. הספרייה `html.parser` (מובנית בפייתון) קוראת את העמוד תג אחר תג. המחלקה `PageFacts` כבר כתובה: היא קוראת עמוד ו**אוספת עובדות** - אל תשנו אותה, רק קראו והריצו את הדוגמה:

</div>

````python
class PageFacts(HTMLParser):
    """Reads an HTML page and collects facts about it. Written for you - do not change."""
    def __init__(self):
        super().__init__()
        self.title = ""        # text inside <title>
        self.h1_texts = []     # the text of every <h1>
        self.images = []       # one dict of attributes per <img>, e.g. {"src": "a.png", "alt": "logo"}
        self.links = []        # one dict of attributes per <a>,   e.g. {"href": "https://..."}
        self.ids = []          # every id="..." on the page
        self._inside = None

    def handle_starttag(self, tag, attrs):
        attrs = dict(attrs)                       # [("href", "x")] -> {"href": "x"}
        if "id" in attrs:
            self.ids.append(attrs["id"])
        if tag == "img":
            self.images.append(attrs)
        if tag == "a":
            self.links.append(attrs)
        if tag == "h1":
            self.h1_texts.append("")
        if tag in ("title", "h1"):
            self._inside = tag

    def handle_endtag(self, tag):
        if tag in ("title", "h1"):
            self._inside = None

    def handle_data(self, data):
        if self._inside == "title":
            self.title += data
        elif self._inside == "h1":
            self.h1_texts[-1] += data

sample = '<title>דוגמה</title><h1>כותרת</h1><img src="a.png"><a href="">ריק</a><p id="x">t</p><!-- <a href="#"> -->'
facts = PageFacts()
facts.feed(sample)
print("title:", facts.title, "| h1:", facts.h1_texts, "| images:", facts.images, "| links:", facts.links, "| ids:", facts.ids)
````

כשקבוצה כותבת מדיניות (בכיתה, בעזרת AI), קל לפספס סעיף, או לכתוב סעיף שהוא רק משפט כללי. נכתוב **בודק מדיניות**: `check_policy(text)` שמוודא שיש את **כל שמונת הסעיפים** ושבכל סעיף יש תוכן, ו-`policy_extras(text)` שבודקת שני דברים שהשיעור הדגיש: **סיווג מידע עם שלושת הצבעים**, ו**סימון נושא משפטי כ"לבדיקת עורך דין"**.

הפונקציה `section_bodies(text)` כבר כתובה: היא מוצאת בטקסט **שורה שמתחילה בשם של סעיף** (גם אם יש לפניה מספר, `#` או `**`) ומחזירה מילון: שם סעיף ← **הטקסט שמתחת לכותרת שלו**, או `None` אם אין כזה סעיף. רשימת הסעיפים (`SECTIONS`, בקיצור) כתובה גם היא. הריצו וראו מה היא מחזירה (שימו לב: גם שארית השורה של הכותרת, כמו `והיקף:**`, נכללת בגוף - זה בסדר):

</div>

````python
SECTIONS = {  # שם קצר של הסעיף -> התבנית שמחפשים בתחילת שורה
    "מטרה": "מטרה",
    "כלים מאושרים": "כלים מאושרים",
    "סיווג מידע": "סיווג מידע",
    "סקירה אנושית": "סקירה אנושית",
    "אבטחה": "אבטחה",
    "קניין רוחני": "קני{1,2}ן רוחני",
    "שקיפות": "שקיפות",
    "אירועים": "אירועים",
}

def section_bodies(text):
    """For each name in SECTIONS: the text under its heading (until the next heading), or None if missing.
    A heading is a line that starts with the name (after an optional number, # or **). Written for you."""
    found = {}
    for name, pattern in SECTIONS.items():
        m = re.search(r"^[^\w\n]*\d*[.)]?[^\w\n]*" + pattern, text, re.MULTILINE)
        if m:
            found[name] = (m.start(), m.end())
    starts = sorted(start for start, end in found.values())
    bodies = {}
    for name in SECTIONS:
        if name not in found:
            bodies[name] = None
        else:
            start, end = found[name]
            later = [s for s in starts if s > start]
            bodies[name] = text[end:(later[0] if later else len(text))]
    return bodies

demo = "1. **מטרה והיקף:** על מי חלה המדיניות\n2. **כלים מאושרים:** רק כלים מהרשימה של הארגון"
for name, body in section_bodies(demo).items():
    print(f"{name}: {body!r}")
````

````python
# בדיקה - חלק ד', שלב 1
FULL_TITLES = ["מטרה והיקף", "כלים מאושרים", "סיווג מידע", "סקירה אנושית", "אבטחה",
               "קניין רוחני ורישיונות", "שקיפות ותיעוד", "אירועים והכשרה"]
FILLER = "כלל קונקרטי אחד שמסביר מי עושה מה ומתי ואיך בודקים זאת בפועל בארגון"
COLORS = "אדום לקוחות וסודות צהוב באישור ירוק נתונים סינתטיים וקוד ציבורי בלבד"

def make_policy(skip=(), style="plain", thin=(), colors=True, lawyer=True):
    lines = []
    for i, title in enumerate(FULL_TITLES, 1):
        if title in skip:
            continue
        body = "קצר" if title in thin else (COLORS if (title == "סיווג מידע" and colors) else FILLER)
        if title == "סיווג מידע" and not colors:
            body = FILLER
        head = {"plain": f"{i}. {title}:", "md": f"## {i}. **{title}**", "bold": f"**{title}:**"}[style]
        lines.append(f"{head} {body}" if style != "md" else f"{head}\n{body}")
    if lawyer:
        lines.append("הערה: כל נושא משפטי דורש בדיקת עורך דין.")
    return "\n".join(lines)

def check_d1():
    for style in ("plain", "md", "bold"):
        assert check_policy(make_policy(style=style)) == [], f"check_policy: מדיניות מלאה (סגנון כותרת {style}) צריכה להחזיר רשימה ריקה"
    assert check_policy(make_policy(skip=("אבטחה", "אירועים והכשרה"))) == ["missing: אבטחה", "missing: אירועים"], \
        "check_policy: שני סעיפים חסרים - missing: <שם קצר מ-SECTIONS>, לפי סדר SECTIONS"
    assert check_policy(make_policy(thin=("סקירה אנושית",))) == ["thin: סקירה אנושית"], "check_policy: סעיף עם פחות מ-8 מילים הוא thin"
    assert check_policy(make_policy().replace("קניין", "קנין")) == [], "check_policy: כתיב ״קנין״ בלי יוד כפולה צריך להתקבל (זה כבר מטופל ב-section_bodies)"
    assert policy_extras(make_policy()) == [], "policy_extras: מדיניות עם צבעים והערת עורך דין - רשימה ריקה"
    assert policy_extras(make_policy(colors=False)) == ["no_colors"], "policy_extras: בסעיף סיווג מידע חסרים הצבעים"
    assert policy_extras(make_policy(lawyer=False)) == ["no_lawyer_note"], "policy_extras: אין התייחסות לעורך דין"
    assert policy_extras(make_policy(skip=("סיווג מידע",))) == [], "policy_extras: אם אין סעיף סיווג - זו בעיה של check_policy, לא של no_colors"
    return True

check_d1() and print("✓ חלק ד' שלב 1: הבודק תופס סעיפים חסרים, דלים, וצבעים והערה משפטית")
````
